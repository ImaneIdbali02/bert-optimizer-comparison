```python
# Installation des dépendances
!pip install transformers datasets wandb torch-optimizer scikit-learn matplotlib seaborn

import os
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, f1_score, classification_report
import time
import gc
import psutil
import wandb
from typing import Dict, List, Tuple, Any

from kaggle_secrets import UserSecretsClient
user_secrets = UserSecretsClient()
secret_value_0 = user_secrets.get_secret("WANDB_API_KEY")
os.environ["WANDB_API_KEY"] = secret_value_0

# Transformers et datasets
from transformers import (
    AutoTokenizer, AutoModelForSequenceClassification,
    TrainingArguments, Trainer, DataCollatorWithPadding,
    Adafactor  # Adafactor est inclus dans transformers
)
from datasets import load_dataset, DatasetDict
import torch_optimizer as optim_extra

# Configuration pour l'accélération GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device utilisé: {device}")

class OptimizationExperiment:
    """Classe pour gérer les expériences de comparaison d'optimiseurs."""
    
    def __init__(self, model_name: str = "bert-base-uncased", max_length: int = 128):
        self.model_name = model_name
        self.max_length = max_length
        self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        self.results = {}
        
    def load_datasets(self) -> Dict[str, DatasetDict]:
        """Charge les datasets GLUE (MRPC et SST-2)."""
        datasets = {}
        
        # MRPC (Microsoft Research Paraphrase Corpus)
        print("Chargement de MRPC...")
        datasets['mrpc'] = load_dataset("glue", "mrpc")
        
        # SST-2 (Stanford Sentiment Treebank)
        print("Chargement de SST-2...")
        datasets['sst2'] = load_dataset("glue", "sst2")
        
        return datasets
    
    def preprocess_function_mrpc(self, examples):
        """Préprocessing pour MRPC (paires de phrases)."""
        return self.tokenizer(
            examples['sentence1'],
            examples['sentence2'],
            truncation=True,
            padding=True,
            max_length=self.max_length
        )
    
    def preprocess_function_sst2(self, examples):
        """Préprocessing pour SST-2 (sentiment analysis)."""
        return self.tokenizer(
            examples['sentence'],
            truncation=True,
            padding=True,
            max_length=self.max_length
        )
    
    def prepare_datasets(self, datasets: Dict[str, DatasetDict]) -> Dict[str, DatasetDict]:
        """Prépare et tokenise les datasets."""
        prepared_datasets = {}
        
        # MRPC
        print("Tokenisation de MRPC...")
        
        # --- CORRECTION ---
        # On garde la colonne 'label' pour la renommer après
        mrpc_cols_to_remove = datasets['mrpc']['train'].column_names
        mrpc_cols_to_remove.remove("label")
        
        mrpc_tokenized = datasets['mrpc'].map(
            self.preprocess_function_mrpc,
            batched=True,
            remove_columns=mrpc_cols_to_remove # On utilise la nouvelle liste
        )
        mrpc_tokenized = mrpc_tokenized.rename_column("label", "labels") # Maintenant, ça fonctionne
        prepared_datasets['mrpc'] = mrpc_tokenized
        
        # SST-2
        print("Tokenisation de SST-2...")

        # --- CORRECTION ---
        # On applique la même logique pour SST-2
        sst2_cols_to_remove = datasets['sst2']['train'].column_names
        sst2_cols_to_remove.remove("label")
        
        sst2_tokenized = datasets['sst2'].map(
            self.preprocess_function_sst2,
            batched=True,
            remove_columns=sst2_cols_to_remove # On utilise la nouvelle liste
        )
        sst2_tokenized = sst2_tokenized.rename_column("label", "labels") # Maintenant, ça fonctionne
        prepared_datasets['sst2'] = sst2_tokenized
        
        return prepared_datasets
    
    def get_optimizer(self, model, optimizer_name: str, learning_rate: float):
        """Crée l'optimiseur selon le type spécifié."""
        if optimizer_name == "AdamW":
            return torch.optim.AdamW(model.parameters(), lr=learning_rate, weight_decay=0.01)
        
        elif optimizer_name == "LAMB":
            return optim_extra.Lamb(model.parameters(), lr=learning_rate, weight_decay=0.01)
        
        elif optimizer_name == "SGD_warmup":
            return torch.optim.SGD(model.parameters(), lr=learning_rate, momentum=0.9, weight_decay=0.01)

        elif optimizer_name == "AdaFactor":
            return Adafactor(
             model.parameters(),
             lr=learning_rate,
             eps=(1e-30, 1e-3), 
             clip_threshold=1.0,
             decay_rate=-0.8,
             beta1=None,
             weight_decay=0.0,
             scale_parameter=True,
             relative_step=False 
            )
        
        else:
            raise ValueError(f"Optimiseur non supporté: {optimizer_name}")
    
    def compute_metrics(self, eval_pred):
        """Calcule les métriques d'évaluation."""
        predictions, labels = eval_pred
        predictions = np.argmax(predictions, axis=1)
        
        accuracy = accuracy_score(labels, predictions)
        f1 = f1_score(labels, predictions, average='weighted')
        
        return {
            'accuracy': accuracy,
            'f1': f1
        }
    
    def monitor_resources(self):
        """Monitore l'utilisation des ressources."""
        # Mémoire GPU
        if torch.cuda.is_available():
            gpu_memory = torch.cuda.memory_allocated() / 1024**3  # GB
            gpu_memory_max = torch.cuda.max_memory_allocated() / 1024**3  # GB
        else:
            gpu_memory = gpu_memory_max = 0
        
        # Mémoire RAM
        ram_usage = psutil.virtual_memory().percent
        
        return {
            'gpu_memory_gb': gpu_memory,
            'gpu_memory_max_gb': gpu_memory_max,
            'ram_usage_percent': ram_usage
        }
    
    def train_with_optimizer(self, dataset_name: str, dataset: DatasetDict, 
                           optimizer_name: str, learning_rate: float = 2e-5,
                           num_epochs: int = 3, batch_size: int = 16) -> Dict[str, Any]:
        """Entraîne le modèle avec un optimiseur spécifique."""
        
        # Initialisation WANDB pour cette expérience
        run_name = f"{dataset_name}_{optimizer_name}_lr{learning_rate}"
        wandb.init(
            project="bert-optimizer-comparison",
            name=run_name,
            config={
                "dataset": dataset_name,
                "optimizer": optimizer_name,
                "learning_rate": learning_rate,
                "epochs": num_epochs,
                "batch_size": batch_size,
                "model": self.model_name
            }
        )
        
        # Chargement du modèle
        num_labels = len(set(dataset['train']['labels']))
        model = AutoModelForSequenceClassification.from_pretrained(
            self.model_name, 
            num_labels=num_labels
        )
        model.to(device)
        
        # Configuration de l'entraînement
        training_args = TrainingArguments(
            output_dir=f"./results_{run_name}",
            num_train_epochs=num_epochs,
            per_device_train_batch_size=batch_size,
            per_device_eval_batch_size=batch_size,
            warmup_steps=500 if optimizer_name == "SGD_warmup" else 0,
            weight_decay=0.01,
            logging_dir=f"./logs_{run_name}",
            logging_steps=50,
            eval_strategy="epoch",  
            save_strategy="no",
            load_best_model_at_end=False,
            metric_for_best_model="accuracy",
            report_to="wandb"
        )
        
        # Préparation des données
        data_collator = DataCollatorWithPadding(tokenizer=self.tokenizer)
        
        # Trainer personnalisé pour utiliser notre optimiseur
        class CustomTrainer(Trainer):
            def create_optimizer(self):
                self.optimizer = self.get_custom_optimizer()
                return self.optimizer
            
            def get_custom_optimizer(self):
                return self.optimizer_func(self.model, optimizer_name, learning_rate)
        
        # Ajout de la fonction d'optimiseur au trainer
        CustomTrainer.optimizer_func = self.get_optimizer
        
        trainer = CustomTrainer(
            model=model,
            args=training_args,
            train_dataset=dataset['train'],
            eval_dataset=dataset['validation'],
            tokenizer=self.tokenizer,
            data_collator=data_collator,
            compute_metrics=self.compute_metrics,
        )
        
        # Monitoring des ressources avant l'entraînement
        start_resources = self.monitor_resources()
        start_time = time.time()
        
        # Reset de la mémoire GPU
        if torch.cuda.is_available():
            torch.cuda.reset_peak_memory_stats()
        
        # Entraînement
        print(f"Début de l'entraînement avec {optimizer_name} sur {dataset_name}...")
        trainer.train()
        
        # Monitoring des ressources après l'entraînement
        end_time = time.time()
        end_resources = self.monitor_resources()
        training_time = end_time - start_time
        
        # Évaluation finale
        eval_results = trainer.evaluate()
        
        # Compilation des résultats
        results = {
            'dataset': dataset_name,
            'optimizer': optimizer_name,
            'learning_rate': learning_rate,
            'training_time': training_time,
            'final_accuracy': eval_results['eval_accuracy'],
            'final_f1': eval_results['eval_f1'],
            'final_loss': eval_results['eval_loss'],
            'gpu_memory_peak_gb': end_resources['gpu_memory_max_gb'],
            'ram_usage_percent': end_resources['ram_usage_percent'],
            'eval_results': eval_results
        }
        
        # Log des résultats dans WANDB
        wandb.log({
            "final_accuracy": results['final_accuracy'],
            "final_f1": results['final_f1'],
            "training_time": results['training_time'],
            "gpu_memory_peak_gb": results['gpu_memory_peak_gb']
        })
        
        wandb.finish()
        
        # Nettoyage de la mémoire
        del model, trainer
        torch.cuda.empty_cache()
        gc.collect()
        
        return results
    
    def run_full_comparison(self, datasets: Dict[str, DatasetDict]) -> pd.DataFrame:
        """Exécute la comparaison complète de tous les optimiseurs."""
        
        optimizers = ["AdamW", "LAMB", "SGD_warmup", "AdaFactor"]
        dataset_names = list(datasets.keys())
        all_results = []
        
        for dataset_name in dataset_names:
            print(f"\n=== DATASET: {dataset_name.upper()} ===")
            
            for optimizer_name in optimizers:
                print(f"\n--- Optimiseur: {optimizer_name} ---")
                
                try:
                    result = self.train_with_optimizer(
                        dataset_name=dataset_name,
                        dataset=datasets[dataset_name],
                        optimizer_name=optimizer_name,
                        learning_rate=2e-5,
                        num_epochs=3,
                        batch_size=16
                    )
                    all_results.append(result)
                    
                    # Sauvegarde intermédiaire
                    temp_df = pd.DataFrame(all_results)
                    temp_df.to_csv(f'intermediate_results.csv', index=False)
                    
                except Exception as e:
                    print(f"Erreur avec {optimizer_name} sur {dataset_name}: {e}")
                    continue
        
        # Création du DataFrame final
        results_df = pd.DataFrame(all_results)
        return results_df
    
    def create_visualizations(self, results_df: pd.DataFrame):
        """Crée des visualisations des résultats."""
        
        # Configuration du style
        plt.style.use('seaborn-v0_8')
        fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        fig.suptitle('Comparaison des Optimiseurs pour Fine-tuning BERT', fontsize=16)
        
        # 1. Accuracy par optimiseur et dataset
        sns.barplot(data=results_df, x='optimizer', y='final_accuracy', hue='dataset', ax=axes[0,0])
        axes[0,0].set_title('Précision par Optimiseur')
        axes[0,0].tick_params(axis='x', rotation=45)
        
        # 2. F1-Score par optimiseur et dataset
        sns.barplot(data=results_df, x='optimizer', y='final_f1', hue='dataset', ax=axes[0,1])
        axes[0,1].set_title('F1-Score par Optimiseur')
        axes[0,1].tick_params(axis='x', rotation=45)
        
        # 3. Temps d'entraînement
        sns.barplot(data=results_df, x='optimizer', y='training_time', hue='dataset', ax=axes[0,2])
        axes[0,2].set_title('Temps d\'Entraînement (secondes)')
        axes[0,2].tick_params(axis='x', rotation=45)
        
        # 4. Utilisation mémoire GPU
        sns.barplot(data=results_df, x='optimizer', y='gpu_memory_peak_gb', hue='dataset', ax=axes[1,0])
        axes[1,0].set_title('Pic Mémoire GPU (GB)')
        axes[1,0].tick_params(axis='x', rotation=45)
        
        # 5. Loss finale
        sns.barplot(data=results_df, x='optimizer', y='final_loss', hue='dataset', ax=axes[1,1])
        axes[1,1].set_title('Loss Finale')
        axes[1,1].tick_params(axis='x', rotation=45)
        
        # 6. Scatter: Accuracy vs Temps d'entraînement
        for dataset in results_df['dataset'].unique():
            subset = results_df[results_df['dataset'] == dataset]
            axes[1,2].scatter(subset['training_time'], subset['final_accuracy'], 
                            label=dataset, s=100, alpha=0.7)
        
        axes[1,2].set_xlabel('Temps d\'Entraînement (s)')
        axes[1,2].set_ylabel('Précision')
        axes[1,2].set_title('Efficacité: Précision vs Temps')
        axes[1,2].legend()
        
        plt.tight_layout()
        plt.savefig('optimizer_comparison_results.png', dpi=300, bbox_inches='tight')
        plt.show()
        
        return fig
    
    def generate_report(self, results_df: pd.DataFrame) -> str:
        """Génère un rapport détaillé des résultats."""
        
        report = """
# RAPPORT DE COMPARAISON DES OPTIMISEURS POUR BERT

## Résumé Exécutif
"""
        
        # Meilleur optimiseur par métrique
        best_accuracy = results_df.loc[results_df['final_accuracy'].idxmax()]
        best_f1 = results_df.loc[results_df['final_f1'].idxmax()]
        fastest = results_df.loc[results_df['training_time'].idxmin()]
        most_efficient_memory = results_df.loc[results_df['gpu_memory_peak_gb'].idxmin()]
        
        report += f"""
### Meilleurs Résultats par Critère:
- **Précision**: {best_accuracy['optimizer']} sur {best_accuracy['dataset']} ({best_accuracy['final_accuracy']:.4f})
- **F1-Score**: {best_f1['optimizer']} sur {best_f1['dataset']} ({best_f1['final_f1']:.4f})
- **Vitesse**: {fastest['optimizer']} sur {fastest['dataset']} ({fastest['training_time']:.2f}s)
- **Mémoire**: {most_efficient_memory['optimizer']} sur {most_efficient_memory['dataset']} ({most_efficient_memory['gpu_memory_peak_gb']:.2f}GB)

## Analyse Détaillée par Dataset
"""
        
        # Analyse par dataset
        for dataset in results_df['dataset'].unique():
            subset = results_df[results_df['dataset'] == dataset]
            report += f"""
### {dataset.upper()}
"""
            report += subset.to_string(index=False)
            report += "\n\n"
        
        # Statistiques globales
        report += """
## Statistiques Globales
"""
        report += results_df.groupby('optimizer').agg({
            'final_accuracy': ['mean', 'std'],
            'final_f1': ['mean', 'std'],
            'training_time': ['mean', 'std'],
            'gpu_memory_peak_gb': ['mean', 'std']
        }).round(4).to_string()
        
        return report

# EXÉCUTION PRINCIPALE
def main():
    """Fonction principale pour exécuter l'expérience complète."""
    
    print(" Début de l'expérience de comparaison des optimiseurs")
    
    # Initialisation
    experiment = OptimizationExperiment()
    
    # Chargement des datasets
    print("\n Chargement des datasets...")
    raw_datasets = experiment.load_datasets()
    
    # Préparation des datasets
    print("\n Préparation et tokenisation...")
    prepared_datasets = experiment.prepare_datasets(raw_datasets)
    
    # Réduction de la taille pour les tests rapides (optionnel)
    # Décommentez les lignes suivantes pour des tests plus rapides
    # for dataset_name in prepared_datasets:
    #     for split in prepared_datasets[dataset_name]:
    #         prepared_datasets[dataset_name][split] = prepared_datasets[dataset_name][split].select(range(100))
    
    # Exécution de la comparaison complète
    print("\n Lancement de la comparaison complète...")
    results_df = experiment.run_full_comparison(prepared_datasets)
    
    # Sauvegarde des résultats
    results_df.to_csv('bert_optimizer_comparison_results.csv', index=False)
    print("\n Résultats sauvegardés dans 'bert_optimizer_comparison_results.csv'")
    
    # Création des visualisations
    print("\n Création des visualisations...")
    experiment.create_visualizations(results_df)
    
    # Génération du rapport
    print("\n Génération du rapport...")
    report = experiment.generate_report(results_df)
    
    with open('optimizer_comparison_report.md', 'w', encoding='utf-8') as f:
        f.write(report)
    
    print("\n Expérience terminée avec succès!")
    print(f" Résultats pour {len(results_df)} configurations testées")
    
    return results_df, report

# Lancement de l'expérience
if __name__ == "__main__":
    results_df, report = main()
    
    # Affichage du résumé
    print("\n" + "="*60)
    print("RÉSUMÉ DES RÉSULTATS")
    print("="*60)
    print(results_df.groupby('optimizer')['final_accuracy'].mean().sort_values(ascending=False))
```
